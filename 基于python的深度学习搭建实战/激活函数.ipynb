```python
import numpy as np
import matplotlib.pyplot as plt
```

# 阶跃函数

```python
def step_function(x):
    # 考虑numpy
    y= x>0
    return y.astype(np.int32)
```

```python
x=np.arange(-5.0,5.0,0.1)
y=step_function(x)
plt.plot(x,y)
```

# sigmoid

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
y=sigmoid(x)
plt.plot(x,y)
```

# ReLU


```python
def ReLU(x):
    return np.maximum(x,0)
```

```python
y=ReLU(x)
plt.plot(x,y)
```

# softmax


```python
def softmax(a):
    a=np.exp(a)
    return a/np.sum(a)
```
