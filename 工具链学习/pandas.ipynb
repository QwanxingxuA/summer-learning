# 绘图

```python
air_quality = pd.read_csv("air_quality_no2.csv",index_col=0,parse_dates=True)
air_quality
```

```python
import matplotlib.pyplot as plt
```

```python
air_quality.plot()
```

```python
air_quality.plot.scatter(x="station_london",y="station_paris")
```

图太多了，具体在使用的时候查手册

```python
air_quality.plot.bar(stacked=True)  #条形图,stacked是否要堆叠
air_quality.plot.barh() #水平条形图
air_quality.plot.hist() #直方图 
air_quality.plot.box() #箱线图
air_quality.plot.area() #面积图
air_quality.plot.hexbin() #6变形填充图
air_quality.plot.pie() #饼图
from pandas.plotting import scatter_matrix
scatter_matrix(air_quality,) #矩阵散点图
air_quality['station_antwerp'].plot.kde() #密度图
```
